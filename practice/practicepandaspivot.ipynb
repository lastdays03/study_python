```python
import pandas as pd
import seaborn as sns
df = sns.load_dataset('titanic')[['age','sex','class','fare','survived']]
df.head()
```

```python
# 선실 등급별로 숙박객의 성별 평균 나이
pdf1 = pd.pivot_table(df,                # 피벗할 데이터 프레임
                      index='class',     # 행 인덱스로 사용
                      columns='sex',     # 컬럼으로 사용
                      values='age',      # 계산 데이터로 사용할 열
                      aggfunc='mean',    # 데이터 집계 함수
                      observed=False     # 경고 해결: 범주형 데이터의 관측되지 않은 값도 포함 (기존 동작 유지) 기본값 False 향 후 True로 변경
                     )
pdf1
```

```python
def peak_to_peak_ratio(x): # x는 시리즈
    # print(x.max())
    # print(x.min())
    return x.max() / x.min() # 함수 반환 값이 수치 스칼라 타입
peak_to_peak_ratio(iris.sepal_length)
```

```python
i_groups.agg(peak_to_peak_ratio)
i_groups.agg(['mean', 'std'])
i_groups.agg({'sepal_length':'mean', 'sepal_width':'std', 'petal_length':'max'})
```

```python
i_groups.agg(lambda x: x.max() / x.min())
```

```python
def top3_petal_length(df):
    print(type(df))
    return df.sort_values(by="petal_length", ascending=False)[:4]
# 함수 반환값이 수치 집합(df가 반환됨)
# type(top3_petal_length(iris))
top3_petal_length(iris)
```

```python
i_groups.apply(top3_petal_length)
```

```python
# i_groups.agg(top3_petal_length)
# agg에 적용된 함수의 반환값이 수치 스칼라가 아니기때문에
# ValueError: Buffer has wrong number of dimensions (expected 1, got 2)
def q3cut(s):
    # print(type(s))
    return pd.qcut(s, 3, labels=["소", "중", "대"]).astype(str)
iris.groupby("species").petal_length.apply(q3cut)
```

```python
# 1. 여러 함수 동시 적용
print(type(df.groupby('부서')))
# print(df.groupby('부서').agg(['sum', 'mean', 'std']))
print(type(df.groupby('부서')['매출']))
print(df.groupby('부서')['매출'].agg(['sum', 'mean', 'std']))
print(type(df.groupby('부서')[['매출', '비용']]))
print(df.groupby('부서')[['매출', '비용']].agg(['sum', 'mean', 'std']))

# 2. 열마다 다른 함수 적용 (딕셔너리 활용)
agg_dict = {
    '매출': 'sum',       # 매출은 합계
    '비용': ['mean', 'max'] # 비용은 평균과 최대값
}
print(df.groupby('부서').agg(agg_dict))
```

```python
def range_func(x):
    """최대값과 최소값의 차이(범위) 계산"""
    return x.max() - x.min()

print(df.groupby('부서')['매출'].agg(range_func))
print(df.groupby('부서')[['매출', '비용']].agg(range_func))
```

```python
result = df.groupby('부서').agg(
    총매출=('매출', 'sum'),
    평균비용=('비용', 'mean'),
    최대매출=('매출', 'max')
)
print(result)
```

```python
# 부서별 매출 평균 계산
dept_mean = df.groupby('부서')['매출'].transform('mean')

# 원본 데이터에 '부서평균매출' 열 추가
df['부서평균매출'] = dept_mean
df['평균대비차이'] = df['매출'] - df['부서평균매출']
df
```

```python
# 매출 합계가 200 이상인 부서만 남기기
def filter_func(x):
    return x['매출'].sum() >= 250

filtered_df = df.groupby('부서').filter(filter_func)
print(filtered_df['부서'].unique())  # 매출 합계가 조건에 맞는 부서만 출력
```

```python
filtered_df2 = df.groupby('부서').filter(lambda x: x['비용'].max() > 50)
filtered_df2
print(filtered_df2['부서'].unique())  # 매출 합계가 조건에 맞는 부서만 출력
```

```python
def top_2_by_sales(x):
    """매출 기준 상위 2개 행 반환"""
    return x.sort_values('매출', ascending=False).head(2)
df.groupby('부서').apply(top_2_by_sales, include_groups=False)
```
