```python
# import asyncio
from playwright.async_api import async_playwright

article_list = []
async def run_async_playwright(url, index):
    # 1. async_playwright 사용
    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=True)
        page = await browser.new_page()
        
        # 2. 모든 Playwright 함수 앞에 await 붙이기
        url = url + str(index)
        await page.goto(url)
        titles = page.locator('.sa_text_strong')
        print(titles)
        
        # for title in titles:
        #     article_list.append(title.text_content())

        await browser.close()

# 3. 비동기 함수 실행
# Jupyter/Colab 환경에서는 asyncio.run() 대신 await을 사용하여 함수를 바로 호출합니다.
# (대부분의 Jupyter 환경에서는 셀의 마지막 await 호출을 자동으로 처리합니다.)

# 일반 파이썬 파일에서는 이렇게 실행해야 합니다:
# if __name__ == "__main__":
#     asyncio.run(run_async_playwright("https://www.example.com"))

# Jupyter/Colab에서는 셀에 바로 아래 코드를 실행:
for i in range(0, 1):
    print(i)
    await run_async_playwright("https://news.naver.com/section/10", i)
```
