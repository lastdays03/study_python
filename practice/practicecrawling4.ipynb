```python
import urllib.request as req
```

```python
# 파일 URL
img_url = "https://sesac.seoul.kr/static/common/images/www/common/img_ssac_intro.png"
html_url = "http://google.com"
```

```python
# 다운받을 경로
save_path1 = "./Myexam/test1.jpg"
save_path2 = "./Myexam/index.html"
```

```python
# urlretrieve 함수는 URL을 통해 파일을 다운로드하고 로컬 파일 시스템에 저장합니다.
# 첫 번째 인자는 다운로드할 파일의 URL이고, 두 번째 인자는 저장될 로컬 파일 경로입니다.
# 함수는 다운로드된 파일의 경로와 HTTPMessage 객체를 반환합니다.
# 예외 처리
try:
    file1, header1 = req.urlretrieve(img_url, save_path1)
    file2, header2 = req.urlretrieve(html_url, save_path2)
except Exception as e:
    print("Download failed.")
    print(e)
else:
    # Header 정보 출력
    print(header1)
    print(header2)
```

```python
# 다운로드 파일 정보
print(f"Filename1 {file1}")
print(f"Filename2 {file2}")
```

```python
import urllib.request as req
from urllib.error import URLError, HTTPError
```

```python
# 다운로드 경로 및 파일명
path_list = ["./MyExam/test2.jpg", "./MyExam/index2.html"]
```

```python
# 다운로드 리소스 URL
target_url = ["https://i.pinimg.com/736x/13/77/b6/1377b6c567a937484e1ce227337ba8bd.jpg", "https://www.python.org/"]
```

```python
for i, url in enumerate(target_url):
    # 예외 처리
    try:
        # 웹 수신 정보 읽기
        response = req.urlopen(url)
        # 수신 내용
        contents = response.read()
        print('---------------------------------------------------')
        # 상태 정보 중간 출력
        print(f'Header Info-{i} : {response.info()}')
        print(f'HTTP Status Code : {response.getcode()}')
        print()
        print('---------------------------------------------------')
        # 파일 쓰기
        with open(path_list[i], 'wb') as c:
            c.write(contents)
        # HTTP 에러 발생 시
    except HTTPError as e:
        print("Download failed.")
        print('HTTPError Code : ', e.code)
        # URL 에러 발생 시
    except URLError as e:
        print("Download failed.")
        print('URL Error Reason : ', e.reason)
        # 성공
    else:
        print()
        print("Download Succeed.")
```
