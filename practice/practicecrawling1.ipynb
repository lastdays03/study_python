```python
import webbrowser
url = 'http://www.naver.com'
webbrowser.open(url, new=0)
```

```python
# 참고: 검색어가 URL에 포함될 때 올바르게 인코딩되도록 urllib.parse.quote_plus를 사용합니다.
# 윈도우 환경에서는 웹 브라우저나 OS가 한글(Non-ASCII)을 자동으로 인코딩하여 처리해주는 경우가 많아 깨짐이 없을 수 있지만,
# macOS나 Linux 환경에서는 인코딩되지 않은 한글을 그대로 전달하면 브라우저에서 인식을 못하거나 깨지는 현상이 발생합니다.
# 따라서 OS 간 호환성을 위해 quote_plus를 사용하여 명시적으로 인코딩하는 것이 안전합니다.
from urllib.parse import quote_plus
naver_search_url = "http://search.naver.com/search.naver?query="
search_word = '파이썬'
encoded_search_word = quote_plus(search_word)
url = naver_search_url + encoded_search_word
webbrowser.open_new(url)
```

```python
# 참고: URL에 프로토콜(http://, https://)을 명시해야 하는 이유
# 윈도우: 'www.naver.com'만 입력해도 OS가 자동으로 'http://'를 붙여 웹 주소로 인식하고 브라우저를 실행해 줍니다.
# macOS/Linux: 프로토콜이 없으면 이를 '로컬 파일 경로'로 인식하여 파일을 찾으려 시도합니다. 해당 파일이 없으므로 브라우저가 열리지 않거나 에러가 발생합니다.
# 결론: OS 간 호환성을 위해 항상 'http://' 또는 'https://'를 명시적으로 붙여주는 것이 안전합니다.
# urls = ['www.naver.com', 'www.daum.net', 'www.google.com']
urls = ['http://www.naver.com', 'http://www.daum.net', 'http://www.google.com']
for url in urls:
    if not url.startswith(('http://', 'https://')):
        url = 'http://' + url
    webbrowser.open_new(url)
```

```python
import requests
from bs4 import BeautifulSoup
import os

# URL에서 이미지 주소 추출
def get_image_urls(url):
    try:
        response = requests.get(url)
        response.raise_for_status()  # HTTP 오류 발생 시 예외 발생
    except requests.exceptions.RequestException as e:
        print(f"URL 접속 오류: {e}")
        return []

    soup = BeautifulSoup(response.text, "lxml")
    image_elements = soup.select('img')
    image_urls = []
    for image_element in image_elements:
        src = image_element.get('src')
        if src:
            # 상대 경로를 절대 경로로 변환 (필요시)
            if src.startswith('//'):
                src = 'https:' + src
            elif src.startswith('/'):
                # 기본 URL에서 도메인 부분을 추출하여 결합
                from urllib.parse import urljoin
                src = urljoin(url, src)
            image_urls.append(src)
    return image_urls

# 폴더를 지정해 이미지 주소에서 이미지 내려받기
def download_image(img_folder, img_url):
    if not img_url:
        print("내려받을 이미지 주소가 유효하지 않습니다.")
        return

    try:
        response = requests.get(img_url, stream=True)
        response.raise_for_status()  # HTTP 오류 발생 시 예외 발생
    except requests.exceptions.RequestException as e:
        print(f"이미지 다운로드 오류 '{img_url}': {e}")
        return

    os.makedirs(img_folder, exist_ok=True)  # 폴더가 없으면 생성
    
    # 파일명 추출 및 유효성 검사
    filename = os.path.basename(img_url)
    # URL에 쿼리 파라미터가 있는 경우 파일명에서 제거
    if '?' in filename:
        filename = filename.split('?')[0]
    
    # 파일명이 너무 짧거나 유효하지 않은 경우 기본값 설정
    if not filename or len(filename) < 3:
        import uuid
        filename = f"downloaded_image_{uuid.uuid4().hex}.jpg" # 고유한 파일명 생성

    file_path = os.path.join(img_folder, filename)

    try:
        with open(file_path, 'wb') as image_file:
            for chunk in response.iter_content(chunk_size=8192): # 8KB씩 청크로 읽기
                image_file.write(chunk)
        print(f"이미지 파일명: '{filename}'. 내려받기 완료!")
    except IOError as e:
        print(f"파일 저장 오류 '{file_path}': {e}")

# 메인 실행 블록
if __name__ == "__main__":
    reshot_url = 'https://www.reshot.com/search/animal'
    figure_folder = "./Myexam/download" # 이미지를 내려받을 폴더 지정

    print(f"'{reshot_url}'에서 이미지 주소 추출 중...")
    reshot_image_urls = get_image_urls(reshot_url)

    if not reshot_image_urls:
        print("추출된 이미지 주소가 없습니다.")
    else:
        num_of_download_image = 3 # 내려받을 이미지 개수 지정
        # num_of_download_image = len(reshot_image_urls) # 모든 이미지 다운로드 시

        print(f"총 {len(reshot_image_urls)}개의 이미지 주소 중 {min(num_of_download_image, len(reshot_image_urls))}개 다운로드 시작.")
        
        for i in range(min(num_of_download_image, len(reshot_image_urls))):
            download_image(figure_folder, reshot_image_urls[i])
        
        print("================================")
        print("선택한 모든 이미지 내려받기 완료!")
```
