```python
import platform
import matplotlib as mpl
import matplotlib.pyplot as plt
from matplotlib import font_manager, rc

system = platform.system()
if system == "Windows":
    font_path = "C:/Windows/Fonts/malgun.ttf"
elif system == "Darwin":
    font_path = "/System/Library/Fonts/AppleSDGothicNeo.ttc"
elif system == "Linux":
    font_path = "/usr/share/fonts/truetype/nanum/NanumGothic.ttf"
else:
    font_path = None
if font_path:
    font = font_manager.FontProperties(fname=font_path).get_name()
    mpl.rcParams['font.family'] = font
else:
    print("폰트 경로를 확인해주세요.")

mpl.rcParams['axes.unicode_minus'] = False
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import os

print(os.getcwd())

# 데이터프레임 생성
d1 = pd.read_excel("notExercise.xls")
print(d1)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(7, 7))
s1 = d1.loc[(d1['분류'] == '남자') | (d1['분류'] == '여자'), '운동을 할 충분한 시간이 없어서']
s1.plot(kind='pie', autopct='%1.1f%%', ax=axes[0, 0])
s2 = d1.loc[(d1['분류'] == '남자') | (d1['분류'] == '여자'), '함께 운동을 할 사람이 없어서']
s2.plot(kind='pie', autopct='%1.1f%%', ax=axes[0, 1])
s2 = d1.loc[(d1['분류'] == '남자') | (d1['분류'] == '여자'), '운동을 할 만한 장소가 없어서']
s2.plot(kind='pie', autopct='%1.1f%%', ax=axes[1, 0])
s2 = d1.loc[(d1['분류'] == '남자') | (d1['분류'] == '여자'), '운동을 싫어해서']
s2.plot(kind='pie', autopct='%1.1f%%', ax=axes[1, 1])
plt.show()
```
